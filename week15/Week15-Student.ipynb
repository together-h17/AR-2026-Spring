````python
# ===== TODO: 修改為你的案例區域和時間 =====

# 案例 A 馬太鞍: ee.Geometry.Rectangle([121.35, 23.55, 121.55, 23.70])
# 案例 B 小林村: ee.Geometry.Rectangle([120.60, 23.10, 120.75, 23.25])
# 案例 C 凱米/台南: ee.Geometry.Rectangle([120.10, 22.90, 120.40, 23.10])
# 案例 D 日月潭: ee.Geometry.Rectangle([120.85, 23.80, 120.98, 23.92])
# 案例 E 桃園埤塘: ee.Geometry.Rectangle([121.15, 24.90, 121.40, 25.10])

AOI = ee.Geometry.Rectangle([120.10, 22.90, 120.40, 23.10])  # ← 替換
START_DATE = '2024-07-01'
END_DATE   = '2024-08-31'

# --- 查詢各資料集 ---
# l5 = ee.ImageCollection('LANDSAT/LT05/C02/T1_L2') \
#     .filterBounds(AOI).filterDate(START_DATE, '2012-12-31')
l7 = ee.ImageCollection('LANDSAT/LE07/C02/T1_L2') \
    .filterBounds(AOI).filterDate(START_DATE, END_DATE)
l8 = ee.ImageCollection('LANDSAT/LC08/C02/T1_L2') \
    .filterBounds(AOI).filterDate(START_DATE, END_DATE)
l9 = ee.ImageCollection('LANDSAT/LC09/C02/T1_L2') \
    .filterBounds(AOI).filterDate(START_DATE, END_DATE)
s2 = ee.ImageCollection('COPERNICUS/S2_SR_HARMONIZED') \
    .filterBounds(AOI).filterDate(START_DATE, END_DATE) \
    .filter(ee.Filter.lt('CLOUDY_PIXEL_PERCENTAGE', 30))

print('=== 資料可用性查詢結果 ===')
#print(f'Landsat 5 (2000-2012): {l5.size().getInfo()} 張')
print(f'Landsat 7 (2024):  {l7.size().getInfo()} 張')
print(f'Landsat 8 (2024):  {l8.size().getInfo()} 張')
print(f'Landsat 9 (2024):  {l9.size().getInfo()} 張')
print(f'Sentinel-2 (雲量<30%): {s2.size().getInfo()} 張')
print()
print('→ 記錄這些數字到 Part A Step 5 的資料需求表中！')
````
